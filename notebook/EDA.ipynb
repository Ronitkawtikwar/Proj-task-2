```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../dataset/cleaned_sales.csv")

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python

df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['sales'], bins=20)
plt.title("Sales Distribution")
plt.show()
```

```python
df['category'].value_counts().plot(kind='bar')

plt.title("Category Distribution")
plt.show()
```

```python
corr = df.corr(numeric_only=True)

corr
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(corr,
            annot=True,
            cmap='coolwarm')

plt.show()
```

```python
sns.scatterplot(x='year', y='sales', data=df)
plt.show()
```

```python
sns.pairplot(
    df[['sales', 'year', 'month']]
)

plt.show()
```
